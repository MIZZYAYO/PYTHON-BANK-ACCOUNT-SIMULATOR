```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        self.transactions = []

    def deposit(self, amount):
        self.balance += amount
        self.transactions.append(f"Deposited {amount}")


ada_account = BankAccount("Ada", 100)
print(ada_account.owner)
print(ada_account.balance)
print(ada_account.transactions)

ada_account = BankAccount("Ada", 100)
ada_account.deposit(50)

print( ada_account.balance)        # 150
print(ada_account.transactions) # Deposited 50
```
