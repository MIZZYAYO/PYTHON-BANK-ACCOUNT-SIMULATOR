```python
# CLASS for the BANKACCOUNT 
 
import json 

class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner  
        self.balance = balance   # Account Balance 
        self.transactions = []     

    def deposit(self, amount):
        self.balance += amount    # adds to initial balance 
        self.transactions.append(f"Deposited ${amount}")  # shows history in list as a transaction 


    def withdraw(self, amount):
      if amount > self.balance:
        print("Error, Insufficient balance") 
      else:
        self.balance -= amount 
        self.transactions.append(f"Withdrew  ${amount}")

    def to_dict(self):
      return {
        "owner": self.owner,
        "balance": self.balance,
        "transactions": self.transactions
            }


        
# TESTING owner and balance 
# ada_account = BankAccount("Ada", 100)
# print(ada_account.owner)  # shows Ada 
# print(f"${ada_account.balance}")  # Shows $100 
# print(ada_account.transactions)






# CREATING a function to create a new account and add it to the dictionary
# Now this "creating an account" function needs 3 things ....Access to the dictionary to create an account, users account name, then stores that users name 




def create_account( accounts, owner, balance=0):
    if owner in accounts:
        print(f"Error: an account for {owner} already exists")   # if the account already exist return ERROR
    else:
        accounts[owner] = BankAccount(owner, balance)

def save_accounts(accounts, filename="accounts.json"):  # created a new function to save_accounts(converts the file to json)
    data = {}    # creates a dict called data to store the data for accounts 
    for name, account in accounts.items():    # registers the name , and the items belonging to the [name] of account 
        data[name] = account.to_dict()   # puts that in a dictionary 
    
    with open(filename, "w") as f:
        json.dump(data, f)

def load_accounts(filename="accounts.json"):
    accounts = {}
    with open(filename, "r") as f:
        data = json.load(f)
    
    for name, info in data.items():
        account = BankAccount(info["owner"], info["balance"])
        account.transactions = info["transactions"]
        accounts[name] = account
    
    return accounts




# CREATING accounts Dictionary to store information on users 
# Without a dictionary, every account needs its own named variable
try:
    accounts = load_accounts()
except FileNotFoundError:
    accounts = {}



# This loop asks users what they want to do when they enter the bank app simulator 

while True:
    print("1. Create account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Check balance")
    print("5. View history")
    print("6. Exit")
    
    choice = input("Choose an option:")
    
    if choice == "1":
        name = input("Enter account owner name: ").strip().capitalize()
        starting_balance = float(input("Enter starting balance: "))
        create_account(accounts, name, starting_balance)
        print(f"Account created for {name} with ${starting_balance}")

    elif choice == "2":
        name = input("Enter account owner name: ").strip().capitalize()
        if name not in accounts:
            print(f"No account found for {name}")
        else:
            amount_depo = float(input("Enter deposit amount: "))
            accounts[name].deposit(amount_depo)
            print(f"Deposited ${amount_depo} to {name}'s account")

    elif choice == "3":
        name = input("Enter account owner name: ").strip().capitalize()
        if name not in accounts:
            print(f"No account found for {name}")
        else:
            amount_withdraw = float(input("Enter withdrawal amount: "))
            accounts[name].withdraw(amount_withdraw)
            print(f"You withdrew ${amount_withdraw} from {name}'s account")

    elif choice == "4":
        name = input("Enter account owner name: ").strip().capitalize()
        if name not in accounts:
            print(f"No account found for {name}")
        else:
            acc_balance = accounts[name].balance
            print(f"Hi {name}, you currently have ${acc_balance} in your account")

    elif choice == "5":
        name = input("Enter account owner name: ").strip().capitalize()
        if name not in accounts:
            print(f"No account found for {name}")
        else:
            history = accounts[name].transactions
            print(f"Transaction history for {name}: -{history}")

    elif choice == "6":
        save_accounts(accounts)
        print("Goodbye!")
        break

    else:
        print("Invalid option, try again")



  

```
